```python
# Named Entity Recognition and Sequence Labeling
# Objectives: Token classification, NER, sequence tagging, BiLSTM-CRF

import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt

torch.manual_seed(42)
```

```python
# Define a custom function with detailed implementation
## Token-Level Classification
class TokenClassifier(nn.Module):
    def __init__(self, vocab_size, embedding_dim, num_classes):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim)
        self.lstm = nn.LSTM(embedding_dim, 128, bidirectional=True, batch_first=True)
        self.classifier = nn.Linear(256, num_classes)
    
    def forward(self, x):
        embeddings = self.embedding(x)
        lstm_out, _ = self.lstm(embeddings)
        logits = self.classifier(lstm_out)  # Per-token classification
        return logits

print("Token Classification Examples:")
print("- POS Tagging: Noun, Verb, Adjective")
print("- NER: Person, Location, Organization")
print("- Chunking: Noun phrase detection")
print("- Slot filling: Intent extraction")
```

```python
# Define a custom function with detailed implementation
## Named Entity Recognition
class NERModel:
    def __init__(self):
        self.tags = {
            'O': 0,      # Outside entity
            'B-PER': 1,  # Beginning of person
            'I-PER': 2,  # Inside person
            'B-LOC': 3,  # Beginning of location
            'I-LOC': 4,  # Inside location
            'B-ORG': 5,  # Beginning of organization
            'I-ORG': 6   # Inside organization
        }
    
    def example_tagging(self):
        text = "John Smith works at Google in California"
        tokens = text.split()
        tags = ['B-PER', 'I-PER', 'O', 'O', 'B-ORG', 'O', 'B-LOC']
        return list(zip(tokens, tags))

ner = NERModel()
print("\nNER BIO Tagging Scheme:")
# Iterate through batches of data
for word, tag in ner.example_tagging():
    print(f"{word:12} {tag}")
```

## Building the Model

Now we'll define our neural network architecture. Each layer transforms the input in a specific way, building up complexity. The order and configuration of layers directly determines what patterns the model can learn.


```python
# Define a custom function with detailed implementation
## CRF (Conditional Random Fields) Layer
class CRFLayer(nn.Module):
    def __init__(self, num_tags):
        super().__init__()
        self.num_tags = num_tags
# Update model parameters based on computed gradients
        self.transitions = nn.Parameter(torch.randn(num_tags, num_tags))
        self.transitions.data[0, :] = -10000  # No transitions to START
        self.transitions.data[:, 0] = -10000  # No transitions from END
    
    def forward(self, emissions, tags):
        # Score sequences with CRF
        score = 0
# Iterate through batches of data
        for t in range(len(tags)):
            score += emissions[t, tags[t]]
            if t > 0:
                score += self.transitions[tags[t-1], tags[t]]
        return score

# Iterate through batches of data
print("\nBiLSTM-CRF for NER:")
print("- BiLSTM generates emission scores")
print("- CRF models tag transitions")
print("- Enforces valid tag sequences (B-PER → I-PER)")
print("- Better than independent token classification")
```

## Evaluation and Inference

After training, we evaluate the model on new data. This shows how well it generalizes beyond training data. Good evaluation practices reveal if the model is truly learning patterns or just memorizing.


```python
# Define a custom function with detailed implementation
## NER Evaluation
class NERMetrics:
    @staticmethod
    def compute_entity_metrics(y_true, y_pred):
        # Extract entities from tag sequences
        true_entities = NERMetrics._extract_entities(y_true)
        pred_entities = NERMetrics._extract_entities(y_pred)
        
        tp = len(true_entities & pred_entities)
        fp = len(pred_entities - true_entities)
        fn = len(true_entities - pred_entities)
        
        precision = tp / (tp + fp) if (tp + fp) > 0 else 0
        recall = tp / (tp + fn) if (tp + fn) > 0 else 0
        f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
        
        return precision, recall, f1
    
    @staticmethod
    def _extract_entities(tags):
        entities = set()
        # Simplified entity extraction
        return entities

print("\nNER Evaluation Metrics:")
print("- Per-token accuracy")
print("- Per-entity precision/recall/F1")
# Iterate through batches of data
print("- Separate metrics for each entity type")
```
